# Video Compression Library Example

This notebook demonstrates how to use the `sdate.video_compression` library in other notebooks.

```python
# Import the video compression functions
from sdate.video_compression import encode_hevc_grayscale_10bit, decode_hevc_grayscale_10bit
import torch
import numpy as np

print("✅ Video compression library imported successfully!")
```

```python
# Create a sample video tensor for demonstration
# In practice, you would load your actual data
T, H, W = 50, 256, 256  # 50 frames of 256x256 images

# Create synthetic data (replace with your actual data loading)
sample_video = torch.rand(T, H, W, dtype=torch.float32)

print(f"Sample video shape: {sample_video.shape}")
print(f"Data type: {sample_video.dtype}")
print(f"Value range: [{sample_video.min():.3f}, {sample_video.max():.3f}]")
```

```python
# Encode the video using HEVC 10-bit compression
output_file = "example_compressed.mov"

encode_hevc_grayscale_10bit(
    v=sample_video,
    outfile=output_file,
    fps=30,
    cq_hw=85,  # Good quality for hardware encoding
    crf_sw=18  # Good quality for software encoding
)

print(f"✅ Video encoded to {output_file}")
```

```python
# Decode the video back to a tensor
decoded_video = decode_hevc_grayscale_10bit(output_file)

print(f"Decoded video shape: {decoded_video.shape}")
print(f"Decoded data type: {decoded_video.dtype}")
print(f"Decoded value range: [{decoded_video.min():.3f}, {decoded_video.max():.3f}]")
```

```python
# Quality analysis (basic example)
# Calculate MSE and PSNR
mse = torch.mean((sample_video - decoded_video) ** 2)
psnr = 20 * torch.log10(1.0 / torch.sqrt(mse)) if mse > 0 else float('inf')

print(f"\n📊 Quality Metrics:")
print(f"   MSE: {mse:.6f}")
print(f"   PSNR: {psnr:.2f} dB")

# File size comparison
import os
original_size = sample_video.numel() * sample_video.element_size()
compressed_size = os.path.getsize(output_file)
compression_ratio = original_size / compressed_size

print(f"\n💾 Compression Analysis:")
print(f"   Original size: {original_size / 1e6:.2f} MB")
print(f"   Compressed size: {compressed_size / 1e6:.2f} MB")
print(f"   Compression ratio: {compression_ratio:.1f}:1")
print(f"   Space savings: {(1 - compressed_size/original_size)*100:.1f}%")
```

## Usage Notes

- The functions work with PyTorch tensors of shape `(T, H, W)` where T is time/frames
- Input tensors should be `float32` with values in range `[0, 1]`
- The encoder tries hardware acceleration first, then falls back to software encoding
- Quality parameters can be adjusted based on your compression vs. quality needs
- See the main documentation in `sdate/video_compression/README.md` for more details
